```python
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)
parent_dir = os.path.dirname(parent_dir)

data_path = os.path.join(parent_dir, "data", 'raw')
data_path
```

```python
files = [
    os.path.join(data_path, "scraped_results_2to50.csv"),
    os.path.join(data_path, "scraped_results_51to251.csv"),
    os.path.join(data_path, "scraped_results_251to500.csv"),
    os.path.join(data_path, "scraped_results_501to750.csv"),
    os.path.join(data_path, "scraped_results_751to1000.csv")
]

df_list = [pd.read_csv(file) for file in files]
merged_df = pd.concat(df_list, ignore_index=True)
crawling_data_path = os.path.join(data_path, "crawling_data.csv")
merged_df.to_csv(crawling_data_path, index=False)
```

```python
df = pd.read_csv(crawling_data_path)
df.head(5)
```
